## Part 1 · Pick one direction

Four directions, pick one. Each is scoped for two weeks of work; if you finish early, add a stretch goal.

### Option A: New physics
Add **advection** so the problem becomes advection-diffusion:
$\partial_t u + \vec{v}\cdot\nabla u = \alpha \nabla^2 u$.
You now have a **CFL** condition from both terms (labCC Part 4 hint). Use an upwind scheme for the advective flux. Compare stability and accuracy against pure diffusion.

### Option B: New algorithm — implicit solve
Replace explicit Euler with **implicit Euler** (unconditionally stable, but requires solving a linear system each step). Use conjugate gradient with a diagonal preconditioner. Compare wall time and stable dt against the explicit version.

### Option C: Adaptive mesh refinement
Refine the mesh where $|\nabla u|$ is large; coarsen where it's small. Use a simple block-structured AMR (two levels). Compare total FLOPs vs uniform mesh for the same accuracy.

### Option D: Port to a new platform
Pick a platform you don't have easy access to — a Frontier AMD MI250X node, an Aurora Intel PVC, or an Apple Silicon M-series with Metal — and describe the porting effort. What API/toolchain? What broke? What didn't?
